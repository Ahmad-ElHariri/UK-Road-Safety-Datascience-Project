#Conclusion about which variables contributed the most to collision type


To understand what drives severity, we ran permutation importance — for each feature, we shuffle its values randomly, breaking its relationship with the target, and measure how much the model's performance drops. Big drop = important feature.
Top factors: collision scale (# of vehicles and # of casualties) (vehicles_x_casualties), speed conditions (speed_group, speed_x_urban), road type (single carriageway, dual carriageway), point of impact, geographic location (longitude), and time (hour, month).
Important note: these are associations, not causes. The model tells us severity is predicted by these factors, not that they directly cause severe crashes.



#Advice to the UK government based on our results

Advice to the UK government: don't spread road safety efforts equally everywhere. Severe crashes concentrate in specific combinations — certain road types, speed contexts, times, and locations. Target those high-risk combinations: strengthen speed controls on single carriageways, concentrate enforcement during peak-risk hours and months, and identify geographic areas where fatal outcomes cluster. Targeted intervention beats uniform effort.
